```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
classes = ["cat", "dog", "house", "car", "tree", "apple", "fish", "bicycle", "star", "airplane"]
img_arr = []

for cls in classes:
    img_arr.append(np.load(f"data/{cls}.npy"))
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for ax, cls, arr in zip(axes.flat, classes, img_arr):
    ax.imshow(arr[0].reshape(28, 28), cmap="gray")
    ax.set_title(cls)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
