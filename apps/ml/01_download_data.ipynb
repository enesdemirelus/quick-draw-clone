```python
import os
import urllib.request

classes = ["cat", "dog", "house", "car", "tree", "apple", "fish", "bicycle", "star", "airplane"]
os.makedirs("data", exist_ok=True)

for c in classes:
    path = os.path.join("data", f"{c}.npy")
    if not os.path.exists(path):
        urllib.request.urlretrieve(f"https://storage.googleapis.com/quickdraw_dataset/full/numpy_bitmap/{c}.npy", path)
        print(f"downloaded {c}")
```
